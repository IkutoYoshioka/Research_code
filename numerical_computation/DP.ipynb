動的計画法は、Bellman方程式を直接解くアプローチで、定常状態周りを局所的に線形近似し解くアプローチに比べて非線形性や不確実性をそのまま扱える。

### 3. Mathmatical Preliminaries

\begin{align*}
v(k)  = \max_{c,y}[U(c) + \beta v(y)]\\
\text{s.t.} \quad c+y &\le f(k)\\
c,y &\ge 0\\
→　⑴
\end{align*}  
* 関数 $f,U$ は $given$  
* 価値関数 $v$ は未知  
上式⑴を満たす$v$の存在と一意性を示すことがタスク。  
  


古典的な方法...  
* **逐次近似法（method of successive approximations）**  
$v_1 \ne v_0$ として $\{v_n\}$ の流列を再帰的に、$$v_{n+1}(k) = \max_{0\le y \le f(k)}\{U[f(k)-y] + \beta v_n(y)\},\quad {n=0,1,2,...}$$によって定義する。
